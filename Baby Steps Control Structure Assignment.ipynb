#BMI Chart
BMI Range<18.5 --->Underweight
BMI Range<18.5 to 24.9 ---> Normal Weight
BMI Range<25 to 29.9 --->Overweight
BMI Range>30 ---> Obese

```python
BMI=int(input("Enter the BMI Index:"))
```

```python
if(BMI<18.5):
    print("Underweight")
elif(BMI<18.5 to 24.9):
    print("Normal Weight")
elif(BMI<25 to 29.9):
    print("Overweight")
else:
    print("Obese")
```

```python
if(BMI<18.5):
    print("Underweight")
elif(BMI<18.5, BMI<24.9):
    print("Normal Weight")
elif(BMI<25 to 29.9):
    print("Overweight")
else:
    print("Obese")
```
