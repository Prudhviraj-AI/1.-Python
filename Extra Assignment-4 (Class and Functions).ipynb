# Create a function that checks whether the given number is Odd or Even

```python
def OddEven():
        num=int(input("Enter a number:"))
        if((num%2)==0):
            print(f"{num} is Even number")
            message="Even Number"
        else:
            print(f"{num} is Odd number")
            message="Odd Number"
        return message
```

```python
OddEven()
```

NOTES:#f"{num}" is an f-string used to insert the value of num into the output.
An f-string (formatted string literal) is used to insert variables or expressions directly inside a string.

# Create a function that tells eligibility of marriage for male and female according to their age limit like 21 for male and 18 for female

```python
def Eligible():
    gender=input("Your Gender:")
    age=int(input("Your Age:"))
    if gender == "Male" and age >= 21:
        print("ELIGIBLE")
        gender="Male"
    elif gender == "Female" and age >= 18:
        print("ELIGIBLE")
        gender="Female"
    else:
        print("NOT ELIGIBLE")
```

```python
Eligible()
```

 #calculate the percentage of your 10th mark

```python
def percentage():
    sub1=int(input("Subject1="))
    sub2=int(input("Subject2="))
    sub3=int(input("Subject3="))
    sub4=int(input("Subject4="))
    sub5=int(input("Subject5="))
    total=sub1+sub2+sub3+sub4+sub5
    percent=total/5
    print("Total:", total)
    print("percentage:",percent)
```

```python
percentage()
```

#print area and perimeter of triangle using class and functions

```python
class Triangle():
    def calculate(self):
        height=int(input("Height:"))
        breadth=int(input("Breadth:"))
        print("Area formula:(Height*Breadth)/2")
        area=(height*breadth)/2
        print("Area of Triangle:", area)
        h1=int(input("Height1:"))
        h2=int(input("Height2:"))
        breadth2=int(input("Breadth:"))
        print("Perimeter formula:Height1+Height2+Breadth")
        perimeter=h1+h2+breadth2
        print("Perimeter of Triangle:", perimeter)
obj = Triangle()
obj.calculate()
```

#Notes:
Triangle-->Defining a class
calculate(self)-->Defining a fuction inside the class, self represents current object of class
obj = Triangle()-->Creates an object of the Triangle class
obj.calculate()-->Calls the function using the object
