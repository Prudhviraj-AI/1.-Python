# Scenario-2:
A program processes a list of numbers and needs to find the largest value. 
Write logic to identify and return the largest number from a given list.

```python
numbers = [15, 48, 7, 91, 36]
largest = numbers[0]
for i in range(len(numbers)):
    if numbers[i] > largest:
        largest = numbers[i]
print("Largest number:", largest)
```

# Scenario-5:
A text-processing tool reverses a given word or sentence for formatting purposes.
Write logic to take a word or sentence as input and produce its reversed version.

```python
text = input("Enter a word or sentence: ")
reversed_text = ""
for temp in text:
    reversed_text = temp + reversed_text

print("Reversed text:", reversed_text)
```

# Scenario-7:
A retail store offers a 20% discount if a customer’s total order exceeds $100. 
Write logic to calculate the final amount to be paid after applying the discount.

```python
total_amount = int(input("Enter the total order amount: "))
if total_amount > 100:
    discount = total_amount * 0.20
else:
    discount = 0
final_amount = total_amount - discount
print("Final amount to be paid:", final_amount)
```

# Scenario-8:
A banking system processes withdrawal requests and ensures the user has enough balance. 
Write logic to check if a user has enough balance before allowing a withdrawal and update
the remaining balance accordingly.

```python
balance = 5000
withdraw = 2000
if withdraw <= balance:
    balance = balance - withdraw
    print("Withdrawal successful")
    print("Remaining balance:", balance)
else:
    print("Insufficient balance Transaction denied.")
```

# Scenario-9:
A calendar system verifies whether a given year is a leap year based on standard leap year rules. 
Write logic to determine whether a given year is a leap year.

```python
year = int(input("Enter a year: "))
if year % 400 == 0: # Checks if the year is divisible by 400. If yes, it is a leap year.
    print(year, "is a leap year.")
elif year % 100 == 0: # If the year is not divisible by 400, check if it’s divisible by 100.Years divisible by 100 but not 400 are not leap years.
    print(year, "is not a leap year.")
elif year % 4 == 0: # If the year is not divisible by 100, check if it’s divisible by 4.
    print(year, "is a leap year.")
else:
    print(year, "is not a leap year.")
```

# Scenario-10:
A program filters out only even numbers from a given list. 
Write logic to extract and return only the even numbers from a list.

```python
numbers = [12, 17, 24, 37, 44, 57]
even_numbers = []
for num in numbers:
    if num % 2 == 0:       
        even_numbers = even_numbers+[num]
print("Even numbers:", even_numbers)
```
