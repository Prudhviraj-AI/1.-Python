# Scenario-1:
You are developing a banking application that categorizes transactions based on the amount entered.
Write logic to determine whether the amount is positive, negative, or zero.

```python
amount = int(input("Enter the transaction amount: "))
if amount > 0:
    print("This is a positive transaction (deposit).")
elif amount < 0:
    print("This is a negative transaction (withdrawal).")
else:
    print("The transaction amount is zero.")
```

# Scenario-2:
A digital locker requires users to enter a numerical passcode. As part of a security feature, the system checks the sum of the digits of the passcode. Write logic to compute the sum of the digits of a given number.

```python
passcode = input("Enter the numerical passcode: ")
digit_sum = 0
for digit in passcode:
    digit_sum += int(digit)
print("Sum of the digits is:", digit_sum)
```

# Scenario-3:
A mobile payment app uses a simple checksum validation where reversing a transaction ID helps detect fraud. 
Write logic to take a number and return its reverse.

```python
def reverse_number(num):
    reversed_num = ""
    for digit in num:
        reversed_num = digit + reversed_num
    return reversed_num
transaction_id = input("Enter the transaction ID: ")
print("Reversed transaction ID:", reverse_number(transaction_id))
```

# Scenario-4:
In a secure login system, certain features are enabled only for users with prime-numbered user IDs. 
Write logic to check if a given number is prime.

```python
n = int(input("Enter user ID: "))
if n > 1:
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            break # using for stop loop immediately
    else:
        print("Prime")
else:
    print("Not Prime")
```

# Scenario-5:
A scientist is working on permutations and needs to calculate the factorial of numbers frequently. 
Write logic to find the factorial of a given number using recursion.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

number = int(input("Enter a number: "))
print("Factorial is:", factorial(number))
```

# Scenario-6:
A unique lottery system assigns ticket numbers where only Armstrong numbers win the jackpot. 
Write logic to check whether a given number is an Armstrong number.


```python
num = input("Enter number: ")
power = len(num)
total = 0
for d in num:
    total = total + int(d) ** power
if total == int(num):
# Notes: A number where the sum of its digits raised to the power of the number of digits equals the number itself.
# Armstrong number = (digit₁)ⁿ + (digit₂)ⁿ + ... + (digitₙ)ⁿ
# 1³ + 5³ + 3³ = 1 + 125 + 27 = 153
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

# Scenario-7:
A password manager needs to strengthen weak passwords by swapping the first and last characters of user-generated passwords. Write logic to perform this operation on a given string.

```python
password = input("Enter the password: ")
if len(password) > 1:
    first = password[0]
    last = password[-1]
    middle = password[1:-1]
    new_password = last + middle + first
else:
    new_password = password
print("Updated password:", new_password)
```

# Scenario-8:
A low-level networking application requires decimal numbers to be converted into binary format before transmission.Write logic to convert a given decimal number into its binary equivalent.


```python
number = int(input("Enter a decimal number: "))
binary = ""
#Notes --> Stores the binary result
for i in range(number):
    remainder = number % 2
    # Notes --> Finds the binary digit (0 or 1)
    binary = str(remainder) + binary
    # Notes --> Adds the digit to the front of the binary string
    number = number // 2
    # Notes --> Reduces the number
    if number == 0:
        break
        # Notes --> Stops when conversion is complete
print("Binary equivalent:", binary)
```

# Scenario-9:
A text-processing tool helps summarize articles by identifying the most significant words. Write logic to find the longest word in a sentence.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
# Splits the sentence into a list of words using spaces.
longest_word = ""
for word in words:
    if len(word) > len(longest_word):
        longest_word = word
print("Longest word:", longest_word)
```

# Scenario-10:
A plagiarism detection tool compares words from different documents and checks if they are anagrams (same characters but different order). 
Write logic to check whether two given strings are anagrams.

```python
word1 = input("Enter first word: ")
word2 = input("Enter second word: ")

for temp in word1:
    word2 = word2.replace(temp, "", 1)

if word2 == "":
    print("Anagrams")
else:
    print("Not anagrams")
```
