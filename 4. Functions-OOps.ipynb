#Odd or even example coding for fuction with retur statement

```python
def OddEven():
        num=int(input("Enter the number:"))
        if((num%2)==1):
            print("Odd Number")
            message="Odd Number"
        else:
            print("Even Number")
            message="Even Number"
        return message
    
```

```python
OddEven()
```

```python
message=OddEven()
```

```python
if(message=="Even Number"):
    print("The give number is Even")
else:
    print("The given number is odd")
```

#BMI Chart Example

```python
def BMI():
    BMI=int(input("Enter the BMI Index:"))
    if(BMI<18.5):
        print("Underweight")
        message="Underweight"
    elif(BMI<24.9):
        print("Normal")
        message="Normal"
    elif(BMI<29.9):
        print("Overweight")
        message="Overweight"
    else:
        print("Very Overweight")
        message="Very Overweight"
    return message
```

```python
bmi=BMI()
```

```python
bmi
```
