1.Write a Python program that prints numbers from 1 to 5.
Expected Output:
1 2 3 4 5



```python
range(1,6) # --> Generates numbers from 1 to 5 (6 is excluded)
for i in range(1, 6): # --> Loops through each number in the range
    print(i, end=' ') # --> Prints the number followed by a space
```

2.Write a Python program to print the first 5 even numbers. Expected Output:
2 4 6 8 10




```python
range(1,6)  # --> Generates numbers from 1 to 5 (6 is excluded)
for i in range(1, 6):   # --> i*2 converts 1,2,3… to 2,4,6… (even numbers).
    print(i*2, end=" ")  # --> prints numbers on the same line.
```

3.Write a Python program to print the sum of numbers from 1 to 10. 
Expected Output:55





```python
total = 0 # --> start with 0.

for i in range(1, 11): # --> loop from 1 to 10.
    total += i  # --> add the current number to the total.

print("Sum:", total) # --> display the sum.
```

4.Write a Python function to find the factorial of 5. Expected Output:
120




```python
def factorial(n): # --> defines a function.
    result = 1 # --> start factorial with 1.
    for i in range(1, n+1): # --> loop from 1 to n.
        result *= i  # --> multiply result by current number.
    return result # --> returns factorial value.

print("Factorial of 5:", factorial(5))
```

5.Write a Python program to check if a number is prime. Expected Output (for input 7):
7 is a prime number




```python
num = int(input("Enter a number: ")) # --> takes user input.
if num > 1: # --> Only numbers greater than 1 can be prime.
    for i in range(2, num): # --> Loops from 2 to num-1 to check divisibility.
        if num % i == 0: # --> If divisible by any number, it’s not prime and stops the loop.
            print(num, "is not a prime number")
            break
        else:
            print(num, "is a prime number") # --> If no divisors found, it’s prime.
else:
    print(num, "is not a prime number") # --> If the number is 1 or less, it’s not prime.
```
