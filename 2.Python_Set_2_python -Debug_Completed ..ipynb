```python
#2.Find the error in the following Python code:
def my_function():
    x += 5
    print(x)

x = 10
my_function()

# Corrected Code --> x is used before being defined inside the function.
# Variables inside a function are local unless declared global
def my_function():
    global x
    x += 5
    print(x)

x = 10
my_function()
```

```python
#5.
def square(n):
    n * n

print(square(4))

# Corrected Code --> Without return, the function returns None.
def square(n):
    return n * n

print(square(4))
```
