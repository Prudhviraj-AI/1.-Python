# Catagory the people by their age like children, adult, citizen, senior citizen...

```python
age=int(input("Enter the age:"))
if(age<18):
    print("Children")
elif(age<35):
    print("Adult")
elif(age<59):
    print("Citizen")
else:
    print("Senior Citizen")
```
